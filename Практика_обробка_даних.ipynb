Завдання 1
Очистіть дані від викидів, розділіть на тренувальну та тестову чатини

```python
y = df['price(in lakhs)']
X = df.drop(columns='price(in lakhs)')
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% даних йде на тест, 80% — на тренування
    random_state=42,    # Фіксує випадковість для відтворюваності результату
)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Визначення числових та категоріальних колонок
num_features = X_train.select_dtypes(include=['int64', 'float64']).columns
cat_features = X_train.select_dtypes(include=['object', 'category']).columns

# Пайплайн для числових даних
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median'))
])

# Пайплайн для категоріальних даних
cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('encoder', OneHotEncoder(handle_unknown='ignore',sparse_output=False))
])

# Об'єднання в один препроцесор
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_features),
    ('cat', cat_pipeline, cat_features)
])

# Застосування до тренувальних та тестових даних
X_train_prepared = preprocessor.fit_transform(X_train)
X_test_prepared = preprocessor.transform(X_test)
```

ЗАВДАННЯ 3 Створіть остаточну модель та натренуйте її

```python
from sklearn.tree import DecisionTreeRegressor

# Створення фінального пайплайну: препроцесор + модель
model = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', DecisionTreeRegressor(random_state=42))
])

# Тренування моделі на сирих тренувальних даних (X_train),
# оскільки пайплайн сам застосує трансформації
model.fit(X_train, y_train)
```

```python
print("Модель успішно натренована!")

# Перевірка результату на тестових даних
train_score = model.score(X_train, y_train)
test_score = model.score(X_test, y_test)

print(f"R^2 на тренувальних даних: {train_score:.4f}")
print(f"R^2 на тестових даних: {test_score:.4f}")
```

```python
import joblib

# Збереження моделі у файл
filename = 'decision_tree_model.joblib'
joblib.dump(model, filename)

print(f"Модель збережена як {filename}")
```

```python
from sklearn.tree import export_graphviz
import graphviz

# Отримуємо саму модель дерева з пайплайну
tree_model = model.named_steps['regressor']

# Отримуємо назви ознак після препроцесора
feature_names = model.named_steps['preprocessor'].get_feature_names_out()

# Експорт дерева у формат DOT
dot_data = export_graphviz(tree_model,
                           out_file=None,
                           feature_names=feature_names,
                           filled=True,
                           rounded=True,
                           special_characters=True,
                           max_depth=3) # Обмежимо глибину для читабельності PDF

# Створення та збереження PDF
graph = graphviz.Source(dot_data)
graph.render("car_price_tree", format="pdf", cleanup=True)

print("Дерево збережено як car_price_tree.pdf")
```
